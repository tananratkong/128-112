```python
orders = [
    {"id": 2001, "customer": "Anan", "city": "Korat", "product": "Keyboard", "qty": 2, "price": 450},

    {"id": 2002, "customer": "Benja", "city": "Bangkok", "product": "Mouse",    "qty": 5, "price": 180},

    {"id": 2003, "customer": "Chai", "city": "Bangkok", "product": "Monitor",  "qty": 1, "price": 3200},

    {"id": 2004, "customer": "Duang", "city": "Phuket", "product": "Mouse",    "qty": 3, "price": 180},

    {"id": 2005, "customer": "Anan", "city": "Korat", "product": "Monitor",  "qty": 2, "price": 3200},

    {"id": 2006, "customer": "Ekk", "city": "Bangkok", "product": "Keyboard", "qty": 4, "price": 450},

]

sales = []         #list ว่าง
total_price = 0         #ตัวแปรเก็บยอดขายรวมทั้งหมด
for order in orders:          #วนลูปเพื่อนอ่าน order ทีละอัน จนครบ
  order_id = order["id"]
  qty = order["qty"]
  price = order["price"]

  subtotal = qty * price          #คำนวณ vat 7% และราคารวม vat สุทธิ
  vat = subtotal * 0.07
  net_total = subtotal + vat

  sales.append((order_id, vat, net_total))          #เพิ่มข้อมูลเข้า tuple

  total_price += net_total          #ยอดสุทธิแต่ละอันมาบวก เพื่อหายอดขายรวมทั้งหมด

print("Sales Report")

for item in sales:          #แสดงข้อมูลของแต่ละ tuple
  print(item)

print("Total Sales =", total_price)
```
