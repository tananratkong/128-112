```python
#ข้อ ป.1 - ชุดฟังก์ชันตรวจข้อมูล
def is_valid_score(score):
  """ตรวจสอบว่าคะแนนอยู่ในช่วง 0 ถึง 100 หรือไม่ คืนค่าเป็น True หรือ False"""
  return 0 <= score <= 100

def to_grade(score):
  """เปลงคะแนนเป็นเกรด A, B, C หรือ F(ถ้าคะแนนถูกต้องแล้ว)"""
  if score >= 80:
    return "A"
  elif score >= 70:
    return "B"
  elif score >= 60:
    return "C"
  else:
    return "F"

def print_result(student_id, score):
  """แสดงผลรายงานรหัสนักศึกษาและเกรด หรือ แจ้งเตือนหากคะแนนไม่ถูกต้อง"""
  if is_valid_score(score):
    grade = to_grade(score)
    print(f"รหัสนักศึกษา: {student_id}, คะแนน: {score}, เกรด: {grade}")
  else:
    print(f"รหัสนักศึกษา: {student_id}, แจ้งเตือน: คะแนนไม่ถูกต้อง(ต้องอยู่ระหว่าง 0-100)")

student_id = input("ป้อนรหัสนักศึกษา: ")
score = float(input("ป้อนคะแนน: "))

print_result(student_id, score)
```

```python
#ข้อ ป.2 - สถิติแบบไหลผ่าน
def safe_averange(total, count):
  if count <= 0:
    return None
  answer = total / count
  return round(answer, 2)

day1 = float(input("กรอกค่า PM2.5 วันที่ 1: "))
day2 = float(input("กรอกค่า PM2.5 วันที่ 2: "))
day3 = float(input("กรอกค่า PM2.5 วันที่ 3: "))
day4 = float(input("กรอกค่า PM2.5 วันที่ 4: "))
day5 = float(input("กรอกค่า PM2.5 วันที่ 5: "))

total = 0
count = 0

if day1 >= 0:
  total += day1
  count += 1
if day2 >= 0:
  total += day2
  count += 1
if day3 >= 0:
  total += day3
  count += 1
if day4 >= 0:
  total += day4
  count += 1
if day5 >= 0:
  total += day5
  count += 1

result = safe_averange(total, count)

if result == None:
  print("ไม่มีข้อมูลที่ใช้ได้")
else:
  print("ค่าเฉลี่ย PM2.5", result)
```
